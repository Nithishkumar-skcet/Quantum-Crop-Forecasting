```python
import pandas as pd

crop = pd.read_csv("datasets/raw/Crop_Yield_Long_Format.csv")
env = pd.read_csv("datasets/raw/Environmental_Features.csv")

crop["District"] = crop["District"].astype(str).str.strip()
env["District"] = env["District"].astype(str).str.strip()

print("Crop districts:", sorted(crop["District"].unique()))
print("\nEnvironmental districts:", sorted(env["District"].unique()))
```

```python
import re

def clean_district(name):
    if pd.isna(name):
        return name

    name = str(name).strip()

    # Remove leading numbers with or without dot
    # Examples:
    # 10. Kanchipuram -> Kanchipuram
    # 2Ranipet -> Ranipet
    # 3Tiruchirappalli -> Tiruchirappalli
    # 1Kanniyakumari -> Kanniyakumari
    name = re.sub(r'^\d+\.?\s*', '', name)

    # Fix known spelling differences
    mapping = {
        "Thoothukudi": "Tuticorin",
        "The nilgiris": "The Nilgiris",
        "The Nilgiris": "The Nilgiris",
        "Thenkasi": "Tenkasi"
    }

    return mapping.get(name, name)

# Apply cleaning
crop["District"] = crop["District"].apply(clean_district)
env["District"] = env["District"].apply(clean_district)
```

```python
print(sorted(crop["District"].unique()))
print(sorted(env["District"].unique()))
```

```python
final_df = crop.merge(
    env,
    on=["District", "Year"],
    how="left",
    indicator=True
)

print(final_df["_merge"].value_counts())
```

```python
# Convert "2015 - 2016" -> 2015
crop["Year"] = (
    crop["Year"]
    .astype(str)
    .str.extract(r"(\d{4})", expand=False)
)

crop["Year"] = pd.to_numeric(crop["Year"], errors="coerce").astype("Int64")
```

```python
env["Year"] = pd.to_numeric(env["Year"], errors="coerce").astype("Int64")
```
