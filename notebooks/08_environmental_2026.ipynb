```python
import pandas as pd
from pathlib import Path

base = Path("datasets/raw/2026")

rainfall = pd.read_csv(base / "TamilNadu_Rainfall_2026.csv")
avg_temp = pd.read_csv(base / "TamilNadu_Temperature_2026.csv")
max_temp = pd.read_csv(base / "TamilNadu_Maximum_Temperature_2026.csv")
min_temp = pd.read_csv(base / "TamilNadu_Minimum_Temperature_2026.csv")
humidity = pd.read_csv(base / "TamilNadu_Humidity_2026.csv")
ndvi = pd.read_csv(base / "TamilNadu_NDVI_2026.csv")
evi = pd.read_csv(base / "TamilNadu_EVI_2026.csv")
lst = pd.read_csv(base / "TamilNadu_LST_2026.csv")
soil_moisture = pd.read_csv(base / "TamilNadu_Soil_Moisture_2026.csv")
et = pd.read_csv(base / "TamilNadu_ET_2026.csv")
solar = pd.read_csv(base / "TamilNadu_Solar_Radiation_2026.csv")
soc = pd.read_csv(base / "TamilNadu_Soil_Organic_Carbon.csv")
ph = pd.read_csv(base / "TamilNadu_Soil_pH.csv")

print("All files loaded successfully.")
```

```python
datasets = {
    "rainfall": rainfall,
    "avg_temp": avg_temp,
    "max_temp": max_temp,
    "min_temp": min_temp,
    "humidity": humidity,
    "ndvi": ndvi,
    "evi": evi,
    "lst": lst,
    "soil_moisture": soil_moisture,
    "et": et,
    "solar": solar,
    "soc": soc,
    "ph": ph
}

for name, data in datasets.items():
    print(f"{name:15} -> {data.columns.tolist()}")
```

```python
# Keep only the required columns from each dataset

rainfall = rainfall[["District", "Year", "Rainfall_mm"]]
avg_temp = avg_temp[["District", "Year", "Avg_Temperature_C"]]
max_temp = max_temp[["District", "Year", "Max_Temperature_C"]]
min_temp = min_temp[["District", "Year", "Min_Temperature_C"]]
humidity = humidity[["District", "Year", "Relative_Humidity_Percent"]]

ndvi = ndvi[["District", "Year", "NDVI"]]
evi = evi[["District", "Year", "EVI"]]
lst = lst[["District", "Year", "LST_C"]]
soil_moisture = soil_moisture[["District", "Year", "Soil_Moisture"]]
et = et[["District", "Year", "Evapotranspiration_mm"]]
solar = solar[["District", "Year", "Solar_Radiation_MJ"]]

soc = soc[["District", "Soil_Organic_Carbon"]]
ph = ph[["District", "Soil_pH"]]

print("Columns cleaned successfully.")
```

```python
# Start with rainfall
env2026 = rainfall.copy()

# Merge time-dependent environmental features
env2026 = env2026.merge(
    avg_temp,
    on=["District", "Year"],
    how="left"
)

env2026 = env2026.merge(
    max_temp,
    on=["District", "Year"],
    how="left"
)

env2026 = env2026.merge(
    min_temp,
    on=["District", "Year"],
    how="left"
)

env2026 = env2026.merge(
    humidity,
    on=["District", "Year"],
    how="left"
)

env2026 = env2026.merge(
    ndvi,
    on=["District", "Year"],
    how="left"
)

env2026 = env2026.merge(
    evi,
    on=["District", "Year"],
    how="left"
)

env2026 = env2026.merge(
    lst,
    on=["District", "Year"],
    how="left"
)

env2026 = env2026.merge(
    soil_moisture,
    on=["District", "Year"],
    how="left"
)

env2026 = env2026.merge(
    et,
    on=["District", "Year"],
    how="left"
)

env2026 = env2026.merge(
    solar,
    on=["District", "Year"],
    how="left"
)

# Merge static soil properties using District only
env2026 = env2026.merge(
    soc,
    on="District",
    how="left"
)

env2026 = env2026.merge(
    ph,
    on="District",
    how="left"
)

print("Merge completed.")
print("Shape:", env2026.shape)
```

```python
print(env2026.head())
print("\nColumns:")
print(env2026.columns.tolist())

print("\nDistricts:", env2026["District"].nunique())
print("Years:", env2026["Year"].unique())
```

```python
print(
    "Duplicate District-Year rows:",
    env2026.duplicated(["District", "Year"]).sum()
)
```

```python
print("\nMissing values:")
print(env2026.isna().sum())
```

```python
# Add State column
env2026.insert(0, "State", "Tamil Nadu")

# Arrange columns in the same order used by the model
env2026 = env2026[
    [
        "State",
        "District",
        "Year",
        "Rainfall_mm",
        "Avg_Temperature_C",
        "Max_Temperature_C",
        "Min_Temperature_C",
        "Relative_Humidity_Percent",
        "NDVI",
        "EVI",
        "LST_C",
        "Soil_Moisture",
        "Evapotranspiration_mm",
        "Solar_Radiation_MJ",
        "Soil_Organic_Carbon",
        "Soil_pH"
    ]
]

# Save
output_path = "datasets/processed/Environmental_2026.csv"

env2026.to_csv(output_path, index=False)

print("Saved successfully:", output_path)
print("Shape:", env2026.shape)
print(env2026.head())
```
