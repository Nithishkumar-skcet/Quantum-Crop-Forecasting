```python
import joblib
import os

os.makedirs("../models", exist_ok=True)

joblib.dump(scaler, "../models/scaler.pkl")
joblib.dump(label_encoders, "../models/label_encoders.pkl")

print("Scaler and Label Encoders saved successfully!")
```
